```python
#importing the Libraies from pandas, matplotlib & numpy
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#Use the pandas library funtion read the csv data file and store it in the dataset variable
dataset = pd.read_csv('Mall_Customers.csv')
```

```python
dataset
```

```python
#column and row wise segretion and get it using iloc[:,:] function
X = dataset.iloc[:, 3:5].values
```

```python
#importing the Libraies from OPTICS clustering
from sklearn.cluster import OPTICS
```

```python
#Below the Optics parameter using to create a model
op = OPTICS(min_samples=50, xi=0.05, min_cluster_size=0.05)
y_op = op.fit_predict(X)
```

```python
# create dataset name is supervised and assign in my dataset
supervised=dataset
```

```python
supervised
```

```python
#Create a new column cluster_group and assign y_op
supervised['Cluster_group']=y_op
```

```python
supervised
```

```python
#Import the seaborn and and create diagram and plot the value for our supervised dataset
import seaborn as sns
facet = sns.lmplot(data=supervised,x=supervised.columns[3], y=supervised.columns[4],  hue=supervised.columns[5], 
                   fit_reg=False, legend=True, legend_out=True)
plt.show()
```
