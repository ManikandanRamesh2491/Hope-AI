```python
#importing the Libraies from pandas, matplotlib & numpy
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#Use the pandas library funtion read the csv data file and store it in the dataset variable
dataset = pd.read_csv('Mall_Customers.csv')
```

```python
dataset
```

```python
#column and row wise segretion and get it using iloc[:,:] function
X = dataset.iloc[:, 3:4].values
```

```python
#How many cluster show or possibility are there to check using Dendrogram method
import scipy.cluster.hierarchy as sch
dendrogram = sch.dendrogram(sch.linkage(X, method = 'ward'))
plt.title('Dendrogram')
plt.xlabel('Customers')
plt.ylabel('Euclidean distances')
plt.show()
```

```python
#importing the Libraies from Agglomerative cluster
from sklearn.cluster import AgglomerativeClustering
```

```python
#Based on above the plot the cluster value is 5, based on that value below the code add
clusmodel = AgglomerativeClustering(n_clusters = 5)
label = clusmodel.fit_predict(X) #model create
```

```python
# create dataset name is supervised and assign in my dataset
supervised=dataset
```

```python
supervised
```

```python
#Create a new column cluster_group and assign label
supervised['Cluster_group']=label
```
